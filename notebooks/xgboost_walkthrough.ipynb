## 2. Take a look at the data

A **row** represents one example; a **feature** is an input column; the **target** is what we want the model to predict. The breast-cancer data has 569 rows and 30 numeric features. We define malignant as the positive class (`1`) so classification recall describes how many malignant examples were found. The diabetes data has 442 rows and 10 features; its target is a numeric disease-progression measure.

```python
X_classification, y_classification = load_classification_data()
X_regression, y_regression = load_regression_data()

print("Classification feature shape:", X_classification.shape)
print("Classification target counts (0 = benign, 1 = malignant):")
display(y_classification.value_counts().sort_index().rename("count"))
print("Regression feature shape:", X_regression.shape)
display(X_classification.head())
```

## 3. Classification: malignant or benign?

The evaluation function holds out 20% of the rows as **test data**. The models learn from the other 80%; the test rows are kept separate to give a fairer estimate of performance on unseen examples. A scikit-learn **pipeline** applies median imputation and standard scaling before the model. Imputation is included to show a reusable pattern, although these bundled datasets have no missing feature values. Scaling is not usually needed by tree-based XGBoost, but helps demonstrate how preprocessing can travel with a model.

The dummy baseline always predicts the most common training class. Precision, recall, and F1 focus on the malignant positive class; accuracy is the fraction of all predictions that are correct.

```python
classification_results = evaluate_classification()
display(classification_results)
```

## 4. Regression: predict a numeric measure

For regression, the dummy baseline always predicts the average target value from the training data. **RMSE** summarizes typical prediction error in the target's units, and lower is better. **R²** compares the model with a simple mean-based prediction; closer to 1 is generally better, while a negative value means it did worse than predicting the test-set mean.

```python
regression_results = evaluate_regression()
display(regression_results)
```

## 5. Save and interpret the results

Compare each XGBoost row with the baseline for the *same task*. Do not compare classification F1 directly with regression RMSE: they measure different things. Results can vary slightly across library versions. The project deliberately computes its results when you run it instead of promising a score in advance.

```python
all_results = run_and_save_results(PROJECT_ROOT / "outputs" / "test_metrics.csv")
display(all_results)
print("Saved to outputs/test_metrics.csv")
```
